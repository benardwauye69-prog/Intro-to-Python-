```python
#sequence [start:stop:step] is an ordered collection of items that can be accessed by their position(index)
#sequence is the list shopping_list[1:2]
#start: Index where the slice begins (incuded)
#stop: Index where the slice ends (exclude)
#step: Number of steps to skip between items
fruits = ["Dragon Fruit", "Mango", "Orange", "Banana"]
                 #0         #1        #2        #3
print(fruits[0:3])
```

```python
# Full copy: nums [:] duplicates the entire sequence.
fruits = ["Dragon Fruit", "Mango", "Orange", "Banana"]
print(fruits[:])
```

```python
# Reversing: nums [::-1] flips the sequence to [9, 8, 7, 6, 5, 4, 3, 2, 1, 0]
fruits = ["Dragon Fruit", "Mango", "Orange", "Banana"]
print(fruits[::-1])
```

```python
# Negative Steps: nums [8:3:-2] counts backward to return [8, 6, 4] #you are going backwards 1,2,3,4 but then u decide to 4,3,2,1
fruits = ["Dragon Fruit", "Mango", "Orange", "Banana"]
print(fruits[3:0:-2]) #sequence[start:stop:step] i.e 1,2,3,4

my_numbers = [1, 2, 3, 4]
             #0 #1  #2  #3  #ascending or descending
             #ascending - smallest to largest -- we start with the lowest index(lowest value) 0
             #descending - largest to smallest -- we start with the largest index(highest value) 3

print(my_numbers[3:0:-2])
```

```python
#Data Analysis, BI analyst, data engineers, data scientist

def clean_sales_data(input_file, output_file):
  """
  read a sales csv file, cleans the data,
  performs transformation and it saves it into
  a new cleaned csv file
  """
  #read the csv file
  df = pd.read_csv(input_file)

  #print the original data
  print("Original data")
  print(df.head()) #shows the first 5 rows

  #clean column names
  #example Product nme -> product_name
  df.columns = (df.columns).str.strip().str.lower().str.replace(" ", "_") #removes extra space, converts to lowercase, and replaces spaces with

  #remove duplicates
  df = df.drop_duplicates()

  #remove rows with missing values
  important_columns = ["product", "quantity", "price"]
  df = df.dropna(subset=important_columns)

  #convert data types

  #ensure quantity and price are numbers
  df['quantity'] = pd.to_numeric(df['quantity'])
  df['price'] = pd.to_numeric(df['price'])

  #create a new column
  #total_sales = quantity * price
  df['total_sales'] = df['quantity'] * df['price']

  #save the cleaned and transformed data
  df.to_csv(output_file, index=False)

  print("Data cleaned successfully")
  print(f"Cleaned file saved as: {output_file}")

  #returned the cleaned df

  return df
```
