```python
#build a simple calculator that does
#addition, subtraction, multiplication, division

a = int(input("Enter a number a "))
b = int(input("Enter a number b "))

#addition
result_add = a + b
print(f"Addition: {a} + {b} = {result_add}")

#subtraction
result_sub = a - b
print(f"Subtraction: {a} - {b} = {result_sub}")

#multiplication
result_multiply = a * b
print(f"Multiplication: {a} * {b} = {result_multiply}")

#division
if b != 0:
    result_div = a / b
    print(f"Division: {a} / {b} = {result_div}")
else:
    print("Error: Division by zero is not allowed.")
```

```python
# Simple Interest Calculator
principal_str = input("Enter the principal amount: ")
principal = float(principal_str.replace(',', ''))

rate_str = input("Enter the annual interest rate (in percentage): ")
rate = float(rate_str.replace('%', ''))

time = float(input("Enter the time period (in years): "))

# Calculate simple interest
simple_interest = (principal * rate * time) / 100

print(f"The principal amount is: ${principal:.2f}")
print(f"The annual interest rate is: {rate:.2f}%")
print(f"The time period is: {time:.2f} years")
print(f"The simple interest is: ${simple_interest:.2f}")
```

```python
#create a grading system for mathematics

score = int(input("enter your score"))
#88 - A
#78 - A-
#68 - B+
#58 - B-
#48 - C+

if score >= 88:
  print("The grade is A")
elif score >= 78:
  print("The grade is A-")
elif score >= 68:
  print("The grade is B+")
elif score >= 58:
  print("The grade is B-")
elif score >= 48:
  print("The grade is C+")
```

```python
#create a system to show
#baby boomer genaplha genz millenial
age = int(input("Enter your age"))

if age <= 10:
  print("This is genalpha")
elif age <= 20:
  print("This is genz")
elif age <= 30:
  print("This is millenial")
elif age <= 40:
  print("This is a baby boomer")
else:
  print("The generation is undefined")

```

```python
#create a 2027 election voting system
# age < 18 - they should not be allowed to take a voters card
#age >= 18 and they should have a national identity card they should
# otherwise they should not take a voters card
```

```python
# 2027 Voting System

print("=== 2027 Voting Registration System ===")

# Acquiring user's name
name = input("Enter your name: ")

# Acquiring user's age
age = int(input("Enter your age: "))

# Age below 18
if age < 18:
  print(name + ", you are NOT allowed to take a voters card")


# Age 18 or above
else:
  # Ask if the user has national ID

  national_id = input("Do you have a national ID? (yes/no): ")

  if national_id.lower() == "yes":
    print(name + ", you are allowed to take a voters card")
  else:
    print(name + ", you are NOT allowed to take a voters card because you do not have a national ID")

```
