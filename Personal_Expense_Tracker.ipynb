```python
#Personal Expense Tracker

from datetime import date
import csv

#global variables
#List of dictionaries to store all expense records
expenses =[]

budget = 0.0

CSV_FILE = "expenses.csv"

FIELDNAMES = [
    "amount",
    "category",
    "description",
    "date"
]

#validation functions
def get_valid_amount():
  while True:
    try:
      amount = float(input("Enter amount (Ksh): "))
      if amount <= 0:
        print("Amount must be greater than 0.")
      else:
        return amount
    except ValueError:
      print("Invalid input. Please enter a valid number")

def get_valid_description():
  while True:
    desc = input("Enter description: ").strip()

    if desc:
      return desc
      print("Description cannot be empty.")


#Categories available
CATEGORIES = ("Food", "Transport", "Rent", "Entertainment", "Health", "Other")

#CORE FUNCTIONS
#function to add another expense
def add_expense():
    print("\n---Add Expense ---")
    amount = get_valid_amount()

    # Amount input and validation
    while True:
        try:
            amount = float(input("Enter amount: "))
            if amount <= 0:
                print("Amount must be greater than 0.")
            else:
                break
        except ValueError:
            print("Please enter a valid number.")


    # Display Categories
    print("\nSelect a category:")
    for i, category in enumerate(CATEGORIES, 1):
        print(f"{i}. {category}")

    #Category Selection
    while True:
        try:
            choice = int(input("Choose category number: "))
            if 1 <= choice <= len(CATEGORIES):
                category = CATEGORIES[choice - 1]
                break
            else:
                print("Invalid category number")
        except ValueError:
            print("Please enter a valid number.")
    description = get_valid_description()

    #Description input
    while True:
        description = input("Enter description: ").strip()
        if description:
            break
        else:
            print("Description cannot be empty.")

    #Date input
    expense_date = input("Enter date (YYYY-MM-DD) or press Enter for today: ").strip()

    if not expense_date:
        expense_date = str(date.today())

    #Create expense dictionary
    expense = {
        "amount": amount,
        "category": category,
        "description": description,
        "date": expense_date

    }

    #Add expense to list
    expenses.append(expense)

    save_to_csv()
    check_budget()

    print("Expense added successfully")

#Function to calculate total spent
def total_spent():
    return sum(e["amount"] for e in expenses)


#Function to view all expenses
def view_expenses():
    if not expenses:
        print("No expenses recorded yet.")
        return

    print("\n--- All Expenses ---")

    print(f"{'#':<4} {'Date':<12} {'Category':<15} {'Description':<20} {'Amount':>10}")
    print("-" * 70)

    for i, e in enumerate(expenses, 1):
        print(
            f"{i<4}"
            f"{e['date']:<12}"
            f"{e['category']:<15}"
            f"{e['description']:<20}"
            f"{e['amount']:>10.2f}"

        )
        print("-" * 70)
        print(f"{'TOTAL':>55} {total_spent():>10.2f}")
```

```python
#Filter expenses by category

def filter_by_category():
    cat = input("Enter category to filter: ").strip().title()
    filtered = [e for e in expenses if e["category"] == cat]

    if not filtered:
        print(f"No expenses found in that category: {cat}")
        return
    print(f"\nExpenses in {cat} category:")
    print("-" * 50)

    for e in filtered:
      print(
          f"{e['date']} | "
          f"{e['description']} | "
          f"Ksh {e['amount']:.2f}"
      )
```

```python
#Category Summary
def category_summary():
  if not expenses:
    print("No expenses recorded yet.")
    return
  summary = {}
  for e in expenses:
    cat = e["category"]
    summary[cat] = summary.get(cat, 0) + e["amount"]
  print("\n--- Category Summary ---")
  for cat, total in summary.items():
    print(f"{cat:<15} Ksh {total:>10.2f}")
```

```python
#unique categories used
def unique_categories():
  if not expenses:
    print("No expenses recorded yet.")
    return
  cats = {e["category"] for e in expenses}
  print("\nCategories Used: ")
  for cat in cats:
    print(f"- {cat}")

```

```python
#most expensive expense
def most_expensive():
  if not expenses:
    print("No expenses recorded yet.")
    return

  top = max(expenses, key=lambda e: e["amount"])
  print(f"\n---Most Expensive Expense---")
  print(f"Date: {top['date']}")
  print(f"Category: {top['category']}")
  print(f"Description: {top['description']}")
  print(f"Amount: Ksh {top['amount']:.2f}")
```

```python
#updating add expenses
def add_expense():
  print("\n--- Add Expense ---")
  amount = get_valid_amount()

  print("\nSelect a category:")
  for i, category in enumerate(CATEGORIES, 1):
    print(f"{i}. {category}")
  while True:
    try:
      choice = int(input("Choose category number: "))
      if 1 <= choice <= len(CATEGORIES):
        category = CATEGORIES[choice - 1]
        break
        print("Invalid category number.")

    except ValueError:
      print("Please enter a valid number.")
  description = get_valid_description()

  expense_date = input("Enter date (YYYY-MM-DD) or press Enter for today: ").strip()
  if not expense_date:
    expense_date = str(date.today())
  expense = {
      "amount": amount,
      "category": category,
      "description": description,
      "date": expense_date
  }
  expenses.append(expense)
  print("Expense added successfully!")

  save_to_csv()

  check_budget()

  print("Expense Added Successfully!")

```

```python
#adding budget status functions

def budget_status():
  if budget == 0:
    print("No budget has been set.")
    return

  spent = total_spent()
  remaining = budget - spent

  print(f"\n--- Budget Status ---")
  print(f"Budget:    Ksh {budget:.2f}")
  print(f"Spent:     Ksh {spent:.2f}")
  print(f"Remaining: Ksh {remaining:.2f}")

  if remaining < 0:
    print(f"You Are Over Budget by Ksh {abs(remaining):.2f}")
```

```python
def save_to_csv():
  with open(
      CSV_FILE,
      "w", # Changed from "W" to "w"
      newline="",

  )  as f:
   writer =csv.DictWriter(
      f,
      fieldnames=FIELDNAMES
   )
   writer.writeheader()
   writer.writerows(expenses)

   print(
       f"Saved "
       f"{len(expenses)} "
       f"expenses(s)."
   )
```

```python
def load_from_csv():
  global expenses
  try:
    with open(
        CSV_FILE,
        "r"
    ) as f:
      reader = csv.DictReader(f)
      expenses = []
      for row in reader:
        row["amount"] = float(
            row["amount"]
        )
        expenses.append(row)
    print(
        f"Loaded "
        f"{len(expenses)} "
        f"expenses(s)."
    )
  except FileNotFoundError: # Corrected indentation for except block
    print(
        "No saved data found."
        "Starting fresh."
    )
```

```python
#Main menu function
def main_menu():
    while True:
        print("\n==== Personal Expense Tracker====")
        print("1. Add Expense")
        print("2. View All Expenses")
        print("3. View Total Spent")
        print("4. Filter By Category")
        print("5. Category Summary")
        print("6. Show Category Used")
        print("7. Most Expensive Expense")
        print("8. Set Budget")
        print("9. Budget Status")
        print("10. Save Expenses")
        print("11. Load Expenses")
        print("12. Exit")

        choice = input("Choose an option: ")

        if choice == "1":
          add_expense()

        elif choice == "2":
          view_expenses()

        elif choice == "3":
          print(f"\nTotal spent: {total_spent():.2f}")

        elif choice == "4":
          filter_by_category()

        elif choice == "5":
          category_summary()

        elif choice == "6":
          unique_categories()

        elif choice == "7":
          most_expensive()

        elif choice == "8":
          set_budget()

        elif choice == "9":
          budget_status()

        elif choice == "10":
          save_to_csv()

        elif choice == "11":
          load_from_csv()

        elif choice == "12":
          print("Exiting program... Goodbye!")
          break

        else:
          print("Invalid option. Please try again.")

#Start the program
load_from_csv()
main_menu()
```
